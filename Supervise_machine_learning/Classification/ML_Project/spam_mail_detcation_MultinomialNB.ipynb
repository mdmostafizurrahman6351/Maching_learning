```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/Classification/Dataset/spam_dataset.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.rename(columns={"message_content":"message", "is_spam":"spam"}, inplace=True)
df.head()
```

```python
df.isnull().sum()
```

```python
df["spam"].value_counts()
```

```python
x = df.message.values
y = df.spam.values

# train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x,y, test_size=0.2, random_state=42)

#TfidfVectorizer
vc = TfidfVectorizer()
x_train = vc.fit_transform(xtrain)
x_test = vc.transform(xtest)


# model create
model = MultinomialNB()
model.fit(x_train, ytrain)
print("Score: ",model.score(x_train, ytrain))

# model predict
y_pred = model.predict(x_test)
print("accuracy_score: ",accuracy_score(ytest, y_pred))
print("confusion_matrix: ",confusion_matrix(ytest, y_pred))
print("classification_report: ",classification_report(ytest,y_pred))
```
