```python
yimport numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import KFold, cross_val_score
```

```python
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/Classification/Dataset/Iris.csv"
df = pd.read_csv(url)
print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop("Id", axis=1,inplace=True)
```

```python
df.info()
```

```python
x = df.drop("Species", axis=1)
y = df["Species"]

model = KNeighborsClassifier(n_neighbors=50)
model.fit(x,y)
```

```python
# evulation
cv = KFold(n_splits=5, shuffle=True, random_state=42)
score = cross_val_score(model,x,y,cv=cv)
print("Scote: ", score)
print("Mean Score: ", score.mean())
```

```python
for k in range(1,20):
    model = KNeighborsClassifier(n_neighbors=k)
    score = cross_val_score(model,x,y,cv=cv)
    print(f"k: {k}, Score: {score.mean()}")
```

```python
# best n_neighbors
model2 = KNeighborsClassifier(n_neighbors=13)

model2.fit(x,y)

#evulation
cv = KFold(n_splits=5, shuffle=True, random_state=42)
score = cross_val_score(model2,x,y,cv=cv)
print("Scote: ", score)

print("Mean Score: ", score.mean())
```
