```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import OneHotEncoder
```

```python
# data load
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/Classification/Dataset/Titanic_%20data.csv"
df = pd.read_csv(url)
df.head()
```

```python
# feature and tergate
x = df[["Pclass","Sex","Age","SibSp","Parch","Fare","Embarked"]]
y = df["Survived"]

print("x colimns null chack: ", x.isnull().sum())
print("y colimns null chack: ", y.isnull().sum())
```

```python
# null value handel
x.loc[:,"Age"] = x["Age"].fillna(x["Age"].median())
x.loc[:,"Embarked"] = x["Embarked"].fillna(x["Embarked"].mode()[0])
print("x data null value check: ", x.isnull().sum())
```

```python
x.info()
```

```python
# preprocessing

# one hot encoding for Embarked & Sex
encoder = OneHotEncoder(sparse_output=False, drop="first")
enc_data = encoder.fit_transform(x[["Sex","Embarked"]])
enc_cols = encoder.get_feature_names_out(["Sex","Embarked"])

x = x.drop(["Sex","Embarked"], axis=1)
enc_x = pd.DataFrame(enc_data, columns=enc_cols, index=x.index)
x = pd.concat([x, enc_x], axis=1)

x.head()
```

```python
# model selection
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# model create
model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)
```

```python
# model predict
y_pred = model.predict(x_test)

# score
print("score: ", model.score(x_test, y_test))
```

```python
# evulation
print("accuracy score: ", accuracy_score(y_test, y_pred))
print("confusion matrix: \n", confusion_matrix(y_test, y_pred))
print("classification report: \n", classification_report(y_test, y_pred))
```
