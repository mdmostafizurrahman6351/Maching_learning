```python
import pandas as pd
from sklearn.model_selection import train_test_split, KFold, StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import OneHotEncoder
```

```python
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/Classification/Dataset/Titanic_%20data.csv"
df = pd.read_csv(url)
df.head()
```

```python
# Feature and tergate
x = df[["Pclass","Sex","Age","SibSp","Parch","Fare","Embarked"]]
y = df["Survived"]
```

```python
print("x data null value check: \n", x.isnull().sum())
print("y data null value check: \n", y.isnull().sum())
```

```python
# null value handel
x.loc[:,"Age"] = x.loc[:,"Age"].fillna(x.loc[:,"Age"].median())
x.loc[:,"Embarked"] = x.loc[:,"Embarked"].fillna(x.loc[:,"Embarked"].mode()[0])
```

```python
# preprocessing
encoder = OneHotEncoder(sparse_output=False, drop="first")
enc_data = encoder.fit_transform(x[["Embarked","Sex"]])
enc_cols = encoder.get_feature_names_out(["Embarked","Sex"])
enc_x = pd.DataFrame(enc_data, columns=enc_cols)

x = x.drop(["Embarked","Sex"], axis=1)
x = pd.concat([x, enc_x], axis=1)

x.head()
```

using Train Test Split

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=0)
```

```python
# model create
model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)

# predict
y_pred = model.predict(x_test)

# score
print("score: ", model.score(x_test, y_test))
print("accuracy: ", accuracy_score(y_test, y_pred))
print("confusion matrix: \n", confusion_matrix(y_test, y_pred))
print("classification report: \n", classification_report(y_test, y_pred))
```

using kfold Cross Valldation

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
print("crose valldiation score: ", cross_val_score(model, x, y, cv=kf))
print("kf score: ", cross_val_score(model, x, y, cv=kf).mean())
```

using StratifiedKFold

```python
# model create
model = LogisticRegression(max_iter=1000)
model.fit(x, y)
print("score", model.score(x, y))
```

```python
st_kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
print("crose valldiation score: ", cross_val_score(model, x, y, cv=st_kf))
print("kf score: ", cross_val_score(model, x, y, cv=st_kf).mean())
```
