Data load

```python
import pandas as pd
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/LinearRegression/Dataset/insurance.csv"
df = pd.read_csv(url)
df.head()
```

data pteprocissing

```python
# encoding
from sklearn.preprocessing import OneHotEncoder, LabelEncoder

# sex & smoker columns label encoding
lab_enc = LabelEncoder()
df["sex"] = lab_enc.fit_transform(df["sex"])
df["smoker"] = lab_enc.fit_transform(df["smoker"])

# region columns one hot encoding
one_enc = OneHotEncoder(drop="first", sparse_output=False)
one_data = one_enc.fit_transform(df[["region"]])
one_columns = one_enc.get_feature_names_out(["region"])

enc_data = pd.DataFrame(one_data, columns=one_columns)

df = df.drop("region", axis=1)

df = pd.concat([df, enc_data], axis=1)
df.head()
```

train test split

```python

# feature and targate
x = df.drop("charges", axis=1)
y = df["charges"]

from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

model create

```python
from sklearn.linear_model import LinearRegression
# model create
model = LinearRegression()

model.fit(x_train, y_train)

# prediction
y_pred = model.predict(x_test)

#score
model.score(x_test, y_test)
```

model evulation

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

print("r2_score: ", r2_score(y_test, y_pred))
print("mean_squared_error: ", mean_squared_error(y_test, y_pred))
print("mean_absolute_error: ", mean_absolute_error(y_test,y_pred))
```

all code together

```python

# library import
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from sklearn.compose import ColumnTransformer

# data load
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/LinearRegression/Dataset/insurance.csv"
df = pd.read_csv(url)

# feature & target
x = df.drop("charges", axis=1)
y = df["charges"]

# column grouping
binary_col = ["sex","smoker"]
region_col = ["region"]
numeric_col = ["age","bmi","children"]

# proprecessing using ColumnsTransformer
preprocessing = ColumnTransformer(
    transformers=[
        ("binuary", OrdinalEncoder(), binary_col),
        ("region", OneHotEncoder(drop="first"),region_col),
        ("num", StandardScaler(), numeric_col)
        ]
)

# full pipeline
model = Pipeline(steps=[
    ("preprocessing", preprocessing),
    ("regressor", LinearRegression())
    ]
)

# train test split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)


# train the model
model.fit(x_train, y_train)

# predict
y_pred = model.predict(x_test)

# Evaluate
r2 = r2_score(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

print("R2 Score:", r2)
print("RMSE:", rmse)
```
