```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

# data load
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/LinearRegression/Dataset/insurance.csv"
df = pd.read_csv(url)

X = df.drop('charges', axis=1)
y = df['charges']

binary_features    = ['sex', 'smoker']
onehot_features    = ['region']
numerical_features = ['age', 'bmi', 'children']

# -------------------------------
# Column Transformer
# -------------------------------
preprocessor = ColumnTransformer(
    transformers=[
        ('binary',  LabelEncoder(),     binary_features),     # LabelEncoder
        ('onehot',  OneHotEncoder(drop='first', sparse_output=False), onehot_features),
        ('num',     'passthrough',      numerical_features)
    ],
    remainder='drop'
)


# Binary columns manuly codding
X['sex']    = X['sex'].map({'female': 0, 'male': 1})
X['smoker'] = X['smoker'].map({'no': 0, 'yes': 1})

# now ColumnTransformer just one-hot + numerical
preprocessor = ColumnTransformer(
    transformers=[
        ('onehot', OneHotEncoder(drop='first', sparse_output=False), onehot_features),
        ('num',    'passthrough', numerical_features)
    ],
    remainder='drop'
)

# Pipeline
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

# fit
pipeline.fit(X, y)
```
