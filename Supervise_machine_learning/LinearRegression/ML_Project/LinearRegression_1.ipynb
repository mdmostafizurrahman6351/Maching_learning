```python
# libary import
import pandas as pd
from google.colab import files
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/LinearRegression/Dataset/Salary_dataset.cs"
df = pd.read_csv(url)
df.head()
```

```python

# extra columns remove
df = df.drop(columns="Unnamed: 0")
df.head()
```

```python

# data show
plt.scatter(df["YearsExperience"], df["Salary"])
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("Salary Data")
plt.show()
```

```python

# input data separate
x = df.drop("Salary", axis=1)
x.head()
```

```python

# output data separate
y = df["Salary"]
y.head()
```

Linear Regression

```python
x.mean()
```

```python
y.mean()
```

```python

# best fit point
plt.scatter(x.mean(), y.mean(), color="orange", label="Best fit point")
plt.scatter(df["YearsExperience"], df["Salary"])
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("Salary Data")
plt.show()
```

```python
reg = LinearRegression() # model create
```

```python
reg.fit(x, y) # train linearregression model
```

```python
m = reg.coef_
m
```

```python
c = reg.intercept_
c
```

```python
# symple System
# y = m*x+c  , x == input value
new_y = m*1.3+c # if YearsExperience=1.3
new_y
```

```python
# sklearn system
new_y = reg.predict([[1.3]])
new_y
```

```python
y_pred = reg.predict(x)
```

```python
plt.plot(x, y_pred, color="g")
plt.scatter(x.mean(), y.mean(), color="orange", label="Best fit point")
plt.scatter(df["YearsExperience"], df["Salary"])
plt.xlabel("YearsExperience")
plt.ylabel("Salary")
plt.title("Salary Data")
plt.show()
```

Lose and Cost function

```python
df["y_pred"] = y_pred
df.head()
```

```python
df["loss"] = df["Salary"] - df["y_pred"]
df.head()
```

```python
# mse & mae
mse = mean_squared_error(df["Salary"], df["y_pred"])
mse
```

```python
mae = mean_absolute_error(df["Salary"], df["y_pred"])
mae
```

```python
# symple tecnic mae calculate
sum(abs(df["loss"])) / len(x)
```

r2 Score Value / Accuracy for Regression

```python
reg.score(x, y)
```

```python
r2_score(y, y_pred)
```
