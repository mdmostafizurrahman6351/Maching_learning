```python

# library import
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python

# data import
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/LinearRegression/Dataset/car_risk_dataset.csv"
df = pd.read_csv(url)
df.head()
```

```python

# check null value
df.isnull().sum()
```

```python
# null value handling using median() or mean() function
df["car_age"] = df["car_age"].fillna(df["car_age"].median())
df["experience"] = df["experience"].fillna(df["experience"].median())
df["speed"] = df["speed"].fillna(df["speed"].median())

df.isnull().sum()
```

```python
# model create
reg = LinearRegression()

# model fit
reg.fit(df[["car_age", "experience", "speed"]], df["risk"])

# predict
risk = reg.predict([[2,6,140]]) # car_age=2, experience=6, speed=140 then risk = __
risk
```

```python
# simple system
m = reg.coef_
m
```

```python
c = reg.intercept_
c
```

```python

# car_age=2, experience=6, speed=140 then risk = __
risk = 2*m[0] + 6*m[1] + 140*m[2] + c
risk # same value return
```
