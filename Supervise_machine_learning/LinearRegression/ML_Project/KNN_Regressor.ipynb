```python
import pandas as pd
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import KFold, cross_val_score
```

```python
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/LinearRegression/Dataset/Fish.csv"

df = pd.read_csv(url, header=2)
df.head()
```

```python
print(df.shape)
df.isnull().sum()
```

```python
print(df.shape)
df.head()
```

```python
x = df.drop(['Weight', 'Species'], axis=1)
y = df['Weight']
```

```python
model = KNeighborsRegressor(n_neighbors=33)
model.fit(x,y)

# evulation
cv = KFold(n_splits=5, shuffle=True, random_state=42)
score = cross_val_score(model, x, y, cv=cv)
print(score)

print("mean score: ", score.mean())
```

```python
for k in range(1,50):
  model2 = KNeighborsRegressor(n_neighbors=k)
  cv = KFold(n_splits=5, shuffle=True, random_state=42)
  score = cross_val_score(model2, x, y, cv=5)
  print("k: ", k, "score: ", score.mean())
```

```python
import matplotlib.pyplot as plt

k_values = list(range(1,50))
score_values = []

for k in k_values:
    model = KNeighborsRegressor(n_neighbors=k)
    cv = KFold(n_splits=5, shuffle=True, random_state=42)
    scores = cross_val_score(model, x, y, cv=5)
    score_values.append(scores.mean())

plt.plot(k_values, score_values, marker='o')
plt.xlabel("K value")
plt.ylabel("R² Score")
plt.title("KNN Regression: K vs R²")
plt.show()
```

```python
# best n_neighbors
model3 = KNeighborsRegressor(n_neighbors=11)
model3.fit(x,y)

# evulation
cv = KFold(n_splits=5, shuffle=True, random_state=42)
score = cross_val_score(model3, x, y, cv=5)
print(score)

print("mean score: ", score.mean())
```
