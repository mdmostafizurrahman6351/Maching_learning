```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/Supervise_machine_learning/LinearRegression/Dataset/Ecommerce_Sales_Prediction_Dataset.csv"
df = pd.read_csv(url)
df.head()
```

```python
# minmax  scalling
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df["Price"] = scaler.fit_transform(df[["Price"]])
df.head()
```

```python
# standard scalling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df["Price"] = scaler.fit_transform(df[["Price"]])
df.head()
```
