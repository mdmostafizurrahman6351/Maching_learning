```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
```

```python

# data import
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/LinearRegression/Dataset/Ice_cream%20selling%20data.csv"
df = pd.read_csv(url)
df.head()
```

```python

# data separate
x = df[["Temperature (°C)"]]
y = df["Ice Cream Sales (units)"]
```

```python
# data view
plt.scatter(x,y,label="original data")
plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales (units)")
plt.title("Polynomial Data")
plt.legend()
plt.show()
```

```python
# polynomial features
poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x)
x_poly.shape
```

```python
# model create
reg = LinearRegression()
reg.fit(x_poly,y)
```

```python
# prediction
y_pred = reg.predict(x_poly)
```

```python
# score
reg.score(x_poly,y)
```

```python
# model view
plt.scatter(x,y,label="original data")
plt.plot(x, y_pred,label="Best fit line", color="orange")
plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales (units)")
plt.title("Predicted Model")
plt.legend()
plt.show()
```
