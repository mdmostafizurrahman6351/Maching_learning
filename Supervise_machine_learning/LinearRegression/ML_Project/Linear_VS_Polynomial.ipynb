```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

# data genarate

```python
# linear data
x_linear = np.linspace(0,10,50)
y_linear = 2*x_linear + 5 + np.random.randn(50) * 2

# non_linear data
x_non_linear = np.linspace(0,10,50)
y_non_linear = 2 * np.sin(x_non_linear) + np.random.randn(50)
```

```python
# linear
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.scatter(x_linear, y_linear, color="blue", label="Data with Linear Relationship")
plt.plot(x_linear, x_linear * 2 + 5, color="green", label="True Linear Relationship")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Relationship")
plt.legend()

# non linear
plt.subplot(1,2,2)
plt.scatter(x_non_linear, y_non_linear, color="g", label="Data with Non Linear Relationship")
plt.plot(x_non_linear, 2 * np.sin(x_non_linear), color="orange", label="True Non Linear Relationship")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Non Linear Relationship")
plt.legend()

plt.tight_layout()
plt.show()
```

correlation coefficient

```python
linear_correlation = np.corrcoef(x_linear, y_linear)[1,0]
non_linear_correlation = np.corrcoef(x_non_linear, y_non_linear)[1,0]

print("linear Relationship correlation coefficient: ", linear_correlation)
print("Non linear Relationship correlation coefficient: ", non_linear_correlation)
```
