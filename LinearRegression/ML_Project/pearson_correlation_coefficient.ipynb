```python
import pandas as pd
```

```python
data = {
"x": [1,2,3,4,5,6,8,7,7,9,11],
"y": [3,4,5,6,5,7,8,8,9,11,13]
}
data
```

```python
df = pd.DataFrame(data)
df
```

```python

# pearson correlation coefficient
df["x"].corr(df["y"]) # y with respect to x
```
