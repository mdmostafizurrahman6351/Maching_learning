```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.cluster import KMeans
```

```python
url = "https://raw.githubusercontent.com/mdmostafizurrahman6351/Maching_learning/refs/heads/main/unsupervised_machine_learning/Dataset/Mall_Customers.csv"
df = pd.read_csv(url)
print(df.shape)
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop("CustomerID", axis=1, inplace=True)
df.rename(columns={"Gender":"gender","Age":"age","Spending Score (1-100)":"score","Annual Income (k$)":"income"}, inplace=True)
df.head()

df["gender"] = df["gender"].map({"Male":1,"Female":0})
df.head()
```

```python
df.describe()
```

```python
plt.scatter(df["age"], df["income"])
plt.xlabel("Age")
plt.ylabel("Income")
plt.title("Age vs Income")
plt.show()
```

```python
plt.scatter(df["score"], df["income"])
plt.xlabel("Score")
plt.ylabel("Income")
plt.title("Score vs Income")
plt.show()
```

```python
sns.pairplot(df)
plt.show()
```

**Defold Optimizing**

```python
km = KMeans()
km.fit(df[["income","score"]])

```

```python
df["new_cluster"] = km.predict(df[["income","score"]])
df.head()
```

```python
df0 = df[df["new_cluster"]==0]
plt.scatter(df0["income"], df0["score"])
plt.xlabel("Income")
plt.ylabel("Score")
plt.title("Cluster 0")
plt.show()
```

```python
df0 = df[df["new_cluster"]==0]
df1 = df[df["new_cluster"]==1]
df2 = df[df["new_cluster"]==2]
df3 = df[df["new_cluster"]==3]
df4 = df[df["new_cluster"]==4]
df5 = df[df["new_cluster"]==5]
df6 = df[df["new_cluster"]==6]
df7 = df[df["new_cluster"]==7]

plt.figure(figsize=(10,10))

plt.scatter(df0.score, df0.income, color="red", label="Cluster 1")
plt.scatter(df1.score, df1.income, color="blue", label="Cluster 2")
plt.scatter(df2.score, df2.income, color="green", label="Cluster 3")
plt.scatter(df3.score, df3.income, color="yellow", label="Cluster 4")
plt.scatter(df4.score, df4.income, color="orange", label="Cluster 5")
plt.scatter(df5.score, df5.income, color="purple", label="Cluster 6")
plt.scatter(df6.score, df6.income, color="pink", label="Cluster 7")
plt.scatter(df7.score, df7.income, color="black", label="Cluster 8")

plt.xlabel("Score")
plt.ylabel("Income")
plt.title("Income and Score plot")
plt.legend()
plt.show()
```

```python
centroids = km.cluster_centers_
centroids
```

```python
df0 = df[df["new_cluster"]==0]
df1 = df[df["new_cluster"]==1]
df2 = df[df["new_cluster"]==2]
df3 = df[df["new_cluster"]==3]
df4 = df[df["new_cluster"]==4]
df5 = df[df["new_cluster"]==5]
df6 = df[df["new_cluster"]==6]
df7 = df[df["new_cluster"]==7]

plt.figure(figsize=(10,10))

plt.scatter(df0.score, df0.income, color="red", label="Cluster 1")
plt.scatter(df1.score, df1.income, color="blue", label="Cluster 2")
plt.scatter(df2.score, df2.income, color="green", label="Cluster 3")
plt.scatter(df3.score, df3.income, color="yellow", label="Cluster 4")
plt.scatter(df4.score, df4.income, color="orange", label="Cluster 5")
plt.scatter(df5.score, df5.income, color="purple", label="Cluster 6")
plt.scatter(df6.score, df6.income, color="pink", label="Cluster 7")
plt.scatter(df7.score, df7.income, color="black", label="Cluster 8")

plt.scatter(centroids[:,0], centroids[:,1], color="lime", label="Centroids", marker="+", s=200)
plt.xlabel("Score")
plt.ylabel("Income")
plt.title("Income and Score plot")
plt.legend()
plt.show()
```

```python
wcss = []
cluster_range = range(1,15)

for k in cluster_range:
  km2 = KMeans(n_clusters=k)
  km2.fit(df[["income","score"]])
  wcss.append(km2.inertia_) # inertia is the wcss value

wcss
```

```python
plt.plot(cluster_range, wcss, marker="o")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
```

```python
#!pip install kneed
from kneed import KneeLocator
kl = KneeLocator(cluster_range, wcss, curve="convex", direction="decreasing")
kl.plot_knee()
print("Knee Point: ", kl.knee)
```

**After Optimizing**

```python
km3 = KMeans(n_clusters=5)
km3.fit(df[["score", "income"]])
```

```python
df["optimized_cluster"] = km3.predict(df[["score", "income"]])
df.head()
#
```

```python
centroids3 = km3.cluster_centers_
centroids3
```

```python
df0 = df[df["optimized_cluster"]==0]
df1 = df[df["optimized_cluster"]==1]
df2 = df[df["optimized_cluster"]==2]
df3 = df[df["optimized_cluster"]==3]
df4 = df[df["optimized_cluster"]==4]

plt.figure(figsize=(10,10))

plt.scatter(df0.score, df0.income, color="red", label="Cluster 1")
plt.scatter(df1.score, df1.income, color="blue", label="Cluster 2")
plt.scatter(df2.score, df2.income, color="green", label="Cluster 3")
plt.scatter(df3.score, df3.income, color="yellow", label="Cluster 4")
plt.scatter(df4.score, df4.income, color="orange", label="Cluster 5")

plt.scatter(centroids3[:,0], centroids3[:,1], color="lime", label="Centroids", marker="+", s=200)


plt.xlabel("Score")
plt.ylabel("Income")
plt.title("Income and Score plot")
plt.legend()
plt.show()
```
